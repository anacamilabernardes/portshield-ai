```python
from pathlib import Path
import os, sys, json, subprocess
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
os.chdir(ROOT)
sys.path.insert(0, str(ROOT / 'src'))
print('Repository:', ROOT.name)
```

```python
if os.getenv('PORTSHIELD_RUN_SERVING') == '1':
    from portshield_ai.benchmark import benchmark
    from portshield_ai.data import read_rows
    prompts = [r['text'] for r in read_rows('data/synthetic/test.jsonl')[:30]]
    for concurrency in (1,4,8):
        report = await benchmark('http://127.0.0.1:8001/v1', 'portshield', prompts, f'results/runs/bench-c{concurrency}.json', concurrency=concurrency)
        display(report)
else:
    print('Serving benchmark not run; set PORTSHIELD_RUN_SERVING=1 with private vLLM running.')
```
